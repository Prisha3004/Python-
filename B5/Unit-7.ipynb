```python
import requests
from bs4 import BeautifulSoup
```

```python
base_url="https://subslikescript.com/movies?page={}"

movie_title=[]
movie_link=[]

for page in range(1,6):
    url=base_url.format(page)
    res=requests.get(url)
    # print(url)
    soup=BeautifulSoup(res.text,"html.parser")
    movie_con =soup.find("ul",class_="scripts-list")
    # print(movie_con)
    movies=movie_con.find_all('a')
    
    for movie in movies:
        #print(movie)
        title=movie.text.strip()
        movie_title.append(title)
        
        link="https://subslikescript.com"+movie['href']
        movie_link.append(link)
        
    
    
```

```python
import pandas as pd
df=pd.DataFrame({
    "Movie_Title":movie_title,
    "Movie_Link":movie_link
})
df
```

```python
website="Fact-checks _ PolitiFact.html"
file=open(website,encoding="utf-8")
res=file.read()
names=[]

soup=BeautifulSoup(res,"html.parser")
n=soup.find_all('a',class_="m-statement__name")
for i in n:
    # print(i.text.strip())
    names.append(i.text.strip())

dates=[]
d=soup.find_all('div',class_="m-statement__desc")
for i in d:
    d1=i.text.strip()
    date_on=d1.split('stated on')[1]
    da=date_on.split('in')[0]
    dates.append(da)

statements=[]
s=soup.find_all('div',class_="m-statement__quote")
for i in s:
    #print(i.text.strip())
    statements.append(i.text.strip())

    
```

```python
df=pd.DataFrame({
    "Source":names,
    "Date":dates,
    "Statements":statements
})
df
```

```python
website="Certified used Mercedes-Benz for sale.html"
file=open(website,encoding="utf-8")
res=file.read()
names=[]
milegae=[]
price=[]
d_name=[]
review=[]
rating=[]

soup=BeautifulSoup(res,"html.parser")
n=soup.find_all('h2',class_="title")
for i in n:
    # print(i.text.strip())
    names.append(i.text.strip())
    
n=soup.find_all('div',class_="mileage")
for i in n:
    # print(i.text.strip())
    milegae.append(i.text.strip())
    
p=soup.find_all('span',class_="primary-price")
for i in p:
    # print(i.text.strip())
    price.append(i.text.strip())
 
p=soup.find_all('div',class_="dealer-name")
for i in p:
    # print(i.text.strip())
    d_name.append(i.text.strip())   

p=soup.find_all(class_="sds-rating__count")
for i in p:
    # print(i.text.strip())
    rating.append(i.text.strip())

p=soup.find_all(class_="sds-rating__link sds-button-link")
for i in p:
    d1=i.text.strip()
    date_on=d1.split('(')[1]
    da=date_on.split('reviews')[0]
    review.append(da)
print(d_name)
print(rating)
print(review)
```
